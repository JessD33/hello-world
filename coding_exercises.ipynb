## Using Coplilots Prompts 

### Lists

```python
# A list containing several names
names = ["Alfie", "Bailey", "Marley"]

# Say hello to each name in the list
for name in names:
    print("Hello, " + name + "!")
```

### Array

```python
# An array of greetings
greetings = ["Hello", "Hiya", "Heyy"]

# Print the first greeting
print(greetings[0] + ", world!")
```

```python
greetings = ["Hello", "Hiya", "Heyy"]

for greeting in greetings:
    print(greeting + ", world!")
```

### Concept of Dictionaries

```python
# A dictionary containing information about a greeting
greeting = {
    "message": "Hello",
    "target": "world"
}

# Use the dictionary values to create the greeting
print(greeting["message"] + ", " + greeting["target"] + "!")
```
